perform element-wise addition on two(2,3) tensors filled with random values using Tensor flow

```python
import tensorflow as tf

# Create two (2,3) tensors with random values
tensor1 = tf.random.uniform(shape=(2, 3))
tensor2 = tf.random.uniform(shape=(2, 3))

# Element-wise addition
result = tf.add(tensor1, tensor2)

# Display results
print("Tensor 1:\n", tensor1.numpy())
print("\nTensor 2:\n", tensor2.numpy())
print("\nElement-wise Addition Result:\n", result.numpy())
```

perform element-wise addition on two(2,3) tensors filled with random values using numpy

```python
import numpy as np

# Create two (2,3) arrays with random values
array1 = np.random.rand(2, 3)
array2 = np.random.rand(2, 3)

# Element-wise addition
result = np.add(array1, array2)

# Display results
print("Array 1:\n", array1)
print("\nArray 2:\n", array2)
print("\nElement-wise Addition Result:\n", result)
```
